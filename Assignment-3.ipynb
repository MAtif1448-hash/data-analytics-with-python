```python
nums = [int(x) for x in input("Element: ").split()]

if nums:
    last = nums[-1]                      
    for i in range(len(nums) - 1, 0, -1):
        nums[i] = nums[i - 1]            
    nums[0] = last                       

print("Rotated list:", nums)
```

```python
def capitalize_words(sentence):
    result = []
    for w in sentence.split():
        result.append(w.capitalize())
    return " ".join(result)

text = input("Sentence: ")
print(capitalize_words(text))
```

```python
def read_matrix(rows, cols):
    matrix = []
    for i in range(rows):
        row = [int(x) for x in input(f"Row {i + 1}: ").split()]
        matrix.append(row)
    return matrix

r1 = int(input("Rows of A: "))
c1 = int(input("Columns of A: "))
r2 = int(input("Rows of B: "))
c2 = int(input("Columns of B: "))

if c1 != r2:
    print("Multiplication not possible: columns of A must equal rows of B.")
else:
    print("Enter matrix A:")
    A = read_matrix(r1, c1)
    print("Enter matrix B:")
    B = read_matrix(r2, c2)

    
    result = [[0] * c2 for _ in range(r1)]

    for i in range(r1):            
        for j in range(c2):        
            for k in range(c1):    
                result[i][j] += A[i][k] * B[k][j]

    print("Product:")
    for row in result:
        print(*row)
```
