Q-6 # Take the radius of a circle from a variable and calculate its Area and Circumference.

```python
radius = 10
π = 3.14159

# Area
Area = π*radius**2

# Circumference

circumference = 2*π*radius

print("Area = ", Area)
print("Circumference = ", circumference)
```

Q-7 # Take the price of an item and a discount percentage in variables, then calculate and print the discount amount and the final price after discount

```python
book_price = 1800
discount_percentage = 5

discounted_amount = (book_price * discount_percentage)/100
final_price = book_price - discounted_amount

print("Final Price = ", int(final_price))
```
